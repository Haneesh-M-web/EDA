```python
import numpy as np
import pandas as pd

data = np.array([
    (1, 'Banuraj', 85),
    (2, 'Rahul', 90),
    (3, 'Kiran', 78),
    (4, 'Arun', 88)
], dtype=[
    ('ID', 'i4'),
    ('Name', 'U10'),
    ('Marks', 'i4')
])

df = pd.DataFrame(data)

print(df)
```

```python
import numpy as np
import pandas as pd

ind=pd.Index([2,3,5,7,11])
print(ind)

print(ind.size,ind.shape,ind.ndim,ind.dtype)


print(ind[1])
print(ind[::2])
print(ind[1:4:2])

indA=pd.Index([1,3,5,7,9])
indB=pd.Index([2,3,5,7,11])

print(indA & indB)
print(indA | indB)
print(indA ^ indB)
print(indA.intersection(indB))

```
